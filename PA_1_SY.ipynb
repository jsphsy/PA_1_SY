A. WORD ROTATION PROBLEM

```python
# A function named word_rotate() that move the first character of the string to the end while keeping all remaining characters in their original order.
def rotate_word(input_string):
        # checks if the string is empty
    if not input_string:
        return ""
        # input_string[1:0] - grabs from the second character onwards
        # input_string[0] - just grabs the first character
    return input_string[1:] + input_string[0]
```

```python
rotate_word("Python") # ythonP
```

```python
rotate_word("Google") # oogleG
```

```python
rotate_word("Electronics") # lectronicsE
```

```python
rotate_word("ProGraMmiNg") # roGraMmingP
```

B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
    # Convert all letters to lowercase
    first_name_lower = first_name.lower()
    last_name_lower = last_name.lower()

    # Remove all spaces from the first name
    first_name_processed = first_name_lower.replace("", "")

    # Remove all spaces from the last name
    last_name_processed = last_name_lower.replace("", "")

    # Join the processed first and last names using one period (.)
    username = f"{first_name_processed}.{last_name_processed}"
    return username
```

```python
make_username("Joseph", "Sy") # 'joseph.sy'
```

```python
make_username("Sherina", "Padilla") # 'sherina.padilla'
```

```python
make_username("Jeanne", "Sy") # 'jeanne.sy'
```

C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    # Use extended sequence unpacking
    # Requirement: first, *middle, last = items
    first, *middle, last = items

    # Return a new list with first and last elements exchanged
    return [last] + middle + [first]
```

```python
print(swap_bookends(["happy", "sad", "angry"])) # ['angry', 'sad', 'happy']
```

```python
print(swap_bookends([6, 7])) # [7, 6]
```

```python
print(swap_bookends(["a", "e", "i", "o", "u"])) # ['u', 'e', 'i', 'o', 'a']
```
